```python
from typing import Literal


def greeting(name : str) -> str | None:
    if name == '':
        return None
    return "Hello " + name
```

```python
greet = greeting(" ")
print(greet)
```

```python
class User:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name.upper()

user = User("Riya")
print(user.name)
```

```python
class Logger:
    def __init__(self):
        print("Logger initialized")

class Service:
    def __init__(self):
        self.logger = Logger()

service = Service()
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2)
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return self.x == other.x and self.y == other.y
```
